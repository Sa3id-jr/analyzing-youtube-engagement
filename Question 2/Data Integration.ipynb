```python
import pandas as pd
```

```python
df1 = pd.read_csv("Tiktok Data Cleaned.xls")
df1.head()
```

```python
df2 = pd.read_csv("Tiktok Source 2 Cleaned.xls")
df2.head()
```

```python
# Rename df2 columns to match df1
df2.rename(columns={'authorUsername': 'username', 'createTime': 'created_at'}, inplace=True)
```

```python
# Engagement = (likes + comments + shares) / views
df2['engagement_rate'] = (df2['likes'] + df2['comments'] + df2['shares']) / df2['views']

# Remove rows with zero views to avoid division by zero
df2 = df2[df2['views'] > 0]
```

```python
cols_needed = ['username', 'created_at', 'likes', 'comments', 'shares', 'views', 'engagement_rate']

df1_clean = df1[cols_needed]  # df1 already has engagement_rate
df2_clean = df2[cols_needed]
```

```python
df = pd.concat([df1_clean, df2_clean], ignore_index=True)
```

```python
df.head(5)
```

```python
df = df.dropna(subset=['created_at'])
```

```python
df['post_date'] = df['created_at'].str.split(' ').str[0]
# just to takee the date only from column created_at
```

```python
df['posts_per_day'] = df.groupby(['username', 'post_date'])['post_date'].transform('count') # counts how many posts each user made on that date and repeats that number for all rows of that user on that date.
```

```python
df['views_per_video'] = df['views'] / 1
df['likes_per_video'] = df['likes'] / 1
df['comments_per_video'] = df['comments'] / 1
```

```python
print(df[['username', 'created_at', 'post_date', 'posts_per_day', 'engagement_rate']].head(5))
```

```python
# Check posting frequency distribution
df['posts_per_day'].value_counts()
```

```python
avg_engagement = df.groupby('posts_per_day')['engagement_rate'].mean()
print(avg_engagement)
# Groups videos by number of posts per day.

# Computes the average engagement rate for users who post 1, 2, 3… times per day.
# help uss answer the question Does posting frequency affect engagement?
```

```python
df.to_csv('Final Integrated Tiktok Data.csv', index=False)


print("Final Integrated Tiktok Data.csv")
```
