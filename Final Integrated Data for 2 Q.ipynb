```python
import pandas as pd
```

```python
df1 = pd.read_csv("tiktok_cleaned.csv")
df1.head()
```

```python
df2 = pd.read_csv("tiktok_source2_clean_aligned.csv")
df2.head()
```

```python
df2.rename(columns={'authorUsername': 'username'}, inplace=True)
```

```python
df1['created_at'] = pd.to_datetime(df1['created_at'])
df2['createTime'] = pd.to_datetime(df2['createTime'])
```

```python
df2.rename(columns={'createTime': 'created_at'}, inplace=True)
```

```python
df2['engagement_rate'] = (
    (df2['likes'] + df2['comments'] + df2['shares']) / df2['views']
)
```

```python
df2 = df2[df2['views'] > 0]
# handle zero views
```

```python
df1_clean = df1[[
    'username', 'created_at', 'likes',
    'comments', 'shares', 'views', 'engagement_rate'
]]

df2_clean = df2[[
    'username', 'created_at', 'likes',
    'comments', 'shares', 'views', 'engagement_rate'
]]
```

```python
df = pd.concat([df1_clean, df2_clean], ignore_index=True)
```

```python
df
```

```python
df = df.dropna(subset=['created_at']) 
# drop invaliddd dates
```

```python
df = df.copy()
df['post_date'] = df['created_at'].dt.date
```

```python
df
```

```python
df
```

```python
# Count how many videos each user posted that day and add as a new column
df['posts_per_day'] = df.groupby(['username', 'post_date'])['post_date'].transform('count')

# Now each video row has the posting frequency for that day
print(df[['username', 'post_date', 'posts_per_day', 'engagement_rate']].head())
```

```python
df
```

```python
# Average engagement rate by posting frequency
avg_engagement = df.groupby('posts_per_day')['engagement_rate'].mean()
print(avg_engagement)
# so we see  the low engagement_rate for post twice per day 
```
