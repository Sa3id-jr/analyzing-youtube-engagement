```python
import pandas as pd
```

```python
df1 = pd.read_csv("Source 1 CLEANED.csv")
df1.head()
```

```python
df2 = pd.read_csv("Source 2 CLEANED.csv")
df2.head()
```

```python
df2['url'] = 'No URL'
df2['category'] = 'Unknown'

print(df2.head())
```

```python

df2 = df2[df1.columns]


print(df2.columns)
```

```python
df = pd.concat([df1, df2], ignore_index=True)

print(df['duration_seconds'].max())
```

```python

df = df.sort_values(by='url', ascending=False)
df = df.drop_duplicates(subset='title', keep='first').reset_index(drop=True)

```

```python
df.head(10)
```

```python
df['title'].duplicated().sum()
```

```python
df.isna().sum() 
```

```python
df['duration_seconds'] = df['duration_seconds'].fillna(df['duration_seconds'].median())
df['comment_count'] = df['comment_count'].fillna(df['comment_count'].median())
df['like_count'] = df['like_count'].fillna(df['like_count'].median()) 
```

```python
# Save to CSV
df.to_csv('Integrated Data Part 1.csv', index=False)


print("Integrated dataset saved to 'Integrated Data Part 1.csv'")
```
